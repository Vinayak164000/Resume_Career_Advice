```python
import getpass
import os
from langchain.agents import create_agent
from langchain_google_genai import ChatGoogleGenerativeAI
from dataclasses import dataclass
from langchain.agents.structured_output import ToolStrategy
```

```python
if "GOOGLE_API_KEY" not in os.environ:
    os.environ["GOOGLE_API_KEY"] = getpass.getpass("Enter your Google AI API key: ")
```

```python
SYSTEM_PROMPT = """You are an expert Academic Counseelor or Career Advisor.


You will be provided with either the prompt or his/her resume:

- prompt: User will provide his career details in bulleted points
- user_pdf: User will provide his resume in pdf format

Based on the above information, first you have to give him resume score out of 100 and then suggest improvements and projects that he can add it to the resume.
Suggestions should not exceeds more than 150 words
"""
```

```python
model = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0,
    max_tokens=None,
    timeout=None,
    max_retries=2,
    
)
```

```python
@dataclass
class CareerAdvice:
    """Response schema for the agent."""
    punny_response: str
    career_advice: str | None = None
```

```python
@dataclass
class Context:
    """Custom runtime context schema."""
    user_id: str
```

```python
agent = create_agent(
    model=model,
    system_prompt= SYSTEM_PROMPT,
    context_schema=Context,
    response_format=ToolStrategy(CareerAdvice),
)
```

```python
config = {"configurable": {"thread_id": "1"}}
```

```python
response = agent.invoke(
    {"messages": [{"role": "user", "content": "I have an experience of working as SDE for 2 years. I know Python, C++, Java languages and has hand-on experience with Flutter and Django framework."}]},
    config=config,
    context=Context(user_id="1")
)
```

```python
print(response['structured_response'].punny_response)
```

```python
print(response['structured_response'].career_advice)
```
